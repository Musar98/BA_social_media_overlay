```python
import torch
import matplotlib.pyplot as plt
%config InlineBackend.figure_format = 'svg'

saved_path="../data/models/checkpoints-ganalyze-dense/checkpoints-ganalyze-dense-b2-ddp-final.pt"
checkpoint = torch.load(saved_path)

loss_history = checkpoint["loss_history"]

print(f"Loaded {len(loss_history)} loss values")

plt.figure(figsize=(10,5))
plt.plot(loss_history)
plt.xlabel("Training step")
plt.ylabel("Loss")
plt.title("Training loss")
plt.grid(True)
plt.savefig("../data/models/checkpoints-ganalyze-dense/checkpoints-ganalyze-dense-b2-ddp-final.loss.svg", format="svg")
plt.show()
```
