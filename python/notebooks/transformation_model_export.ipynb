# Transformation Model Export
This notebook shows how to export the PyTorch model that just applies the global image transformations using the supplied parameters

```python
from social_media_overlay.models.transformations import Transformations
import torch

transformationModel = Transformations()

transformationModel.eval()

example_inputs = (
    torch.randn(1, 3, 480, 480),             # images
    torch.randn(1, 37, dtype=torch.float32), # params
)

dyn_shapes = {
    "x": {0: "batch", 2: "height", 3: "width"},
    "params": {0: "batch"},
}

torch.onnx.export(
    transformationModel,
    example_inputs,
    "transformations.dyn.onnx",
    input_names=["images", "params"],
    output_names=["output_images"],
    opset_version=21,
    dynamic_shapes=dyn_shapes,
    dynamo=True,
    report=True,
    verify=True,
    external_data=False, # no separate data file!
)
```
